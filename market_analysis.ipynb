```python
!pip install yfinance -q
import yfinance as yf
import pandas as pd

btc = yf.download('BTC-USD', start='2022-01-01', end='2023-12-31', progress=False)
px = btc['Close'].squeeze()

peak = px.cummax()
dd = (px - peak) / peak

trough_date = dd.idxmin()
peak_date = px.loc[:trough_date].idxmax()
months = (trough_date - peak_date).days / 30.44

print('最大回撤      :', round(dd.min()*100, 1), '%')
print('峰值日期/价格 :', peak_date.date(), '$', round(float(px[peak_date])))
print('谷底日期/价格 :', trough_date.date(), '$', round(float(px[trough_date])))
print('历时          :', round(months, 1), '个月')
```

```python
!pip install yfinance -q
import yfinance as yf, pandas as pd

TICKERS = ['NVDA', 'AAPL', 'MSFT', 'SHEL.L', 'HSBA.L']   # ← NVDA确定，其他换成你想测的
START, END = '2025-10-01', '2025-11-05'

px = yf.download(TICKERS + ['^FTSE'], start=START, end=END, progress=False)['Close']
ret = px.iloc[-1] / px.iloc[0] - 1

port  = ret[TICKERS].mean()
bench = ret['^FTSE']

print('组合收益  :', round(port*100, 2), '%')
print('FTSE基准  :', round(bench*100, 2), '%')
print('超额收益  :', round((port-bench)*100, 2), '%')
print()
for t in TICKERS:
    print(f'  {t:10s} {ret[t]*100:+.2f}%')
```

```python
WACC, G = 0.0766, 0.02
FCFF   = [2840.74, 2908.66, 2978.48, 3050.26, 3124.04]   # FY2026–FY2030
FCFF31 = 3199.90                                          # FY2031,终值的基数
CASH, INVEST, ASSOC, DEBT, SHARES = 4629, 958, 110, 19649, 6699.6

pv    = sum(f / (1+WACC)**(i+1) for i, f in enumerate(FCFF))
tv    = FCFF31 * (1+G) / (WACC - G)
pv_tv = tv / (1+WACC)**5
ev    = pv + pv_tv
eq    = ev + INVEST + ASSOC + CASH - DEBT

print(f'PV of FCFFs   : £{pv:,.0f}m')
print(f'Terminal Value: £{tv:,.0f}m')
print(f'PV of TV      : £{pv_tv:,.0f}m')
print(f'企业价值      : £{ev:,.0f}m')
print(f'股权价值      : £{eq:,.0f}m')
print(f'每股内在价值  : {eq/SHARES*100:.1f}p')
```
